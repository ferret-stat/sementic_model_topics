```python
import os
import sys

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
path = os.getcwd().replace("\\notebooks", "")
sys.path.append(path)
```

```python
df = pd.read_csv(path + "\\data\\datasets.csv")
df
```

```python
class_counts = df["sentiment"].value_counts().sort_index()
class_counts
```

```python
df["char_len"] = df["text"].astype(str).apply(len)
df["word_len"] = df["text"].astype(str).apply(lambda x: len(x.split()))
df["char_len"].describe()
```

```python
imbalance_ratio = class_counts.max() / class_counts.min()
imbalance_ratio
```

```python
class_counts.plot(kind='bar')
plt.title("Распределение классов")
plt.xlabel("Класс")
plt.ylabel("Количество")
plt.show()
```
